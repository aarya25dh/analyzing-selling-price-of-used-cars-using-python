```python
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/car_new_details.csv")
```

```python
categorical_columns = df.select_dtypes(include=["object", "string"]).columns

for col in categorical_columns:
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
print("\nModel frequency statistics:")
print(df["Model"].value_counts().describe())
```

```python
print("\nMost common models:")
print(df["Model"].value_counts().head(30))
```

```python
print("\nNumber of models by frequency:")
print(df["Model"].value_counts().value_counts().sort_index())
```

```python
print("Number of models in each Make:")
print(df.groupby("Make")["Model"].nunique().sort_values(ascending=False))
```

```python
model_make_counts = df.groupby("Model")["Make"].nunique()

print("Models associated with multiple Makes:")
print(model_make_counts[model_make_counts > 1])
```

```python
df["Owner"] = df["Owner"].replace({
    "UnRegistered Car": "Unregistered Car"
})

df.to_csv(
    "../data/processed/car_new_details.csv",
    index=False
)
```

```python
fuel_check = df[df["Fuel Type"].isin([
    "CNG + CNG",
    "Petrol + CNG",
    "Petrol + LPG"
])]

print(fuel_check[[
    "Make",
    "Model",
    "Fuel Type",
    "Price",
    "Engine",
    "Max Power",
    "Max Torque"
]])
```

```python
drivetrain_check = df[df["Drivetrain"] == "Unknown"]

print(drivetrain_check[[
    "Make",
    "Model",
    "Price",
    "Engine",
    "Max Power",
    "Max Torque"
]])
```

```python
print(df["Year"].describe())

print("\nMinimum year:", df["Year"].min())
print("Maximum year:", df["Year"].max())

print("\nYear distribution:")
print(df["Year"].value_counts().sort_index())
```
