```python
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/car_new_details.csv")
```

```python
df["Car_Age"] = 2022 - df["Year"]
```

```python
df["Kilometers_Per_Year"] = df["Kilometer"] / df["Car_Age"].replace(0, 1)
```

```python
owner_mapping = {
    "Unregistered Car": 0,
    "First": 1,
    "Second": 2,
    "Third": 3,
    "Fourth": 4,
    "4 or More": 4
}

df["Owner_Count"] = df["Owner"].map(owner_mapping)

df["Unregistered"] = (df["Owner"] == "Unregistered Car").astype(int)
```

```python
df[[
    "Year",
    "Car_Age",
    "Kilometer",
    "Kilometers_Per_Year",
    "Owner",
    "Owner_Count",
    "Unregistered"
]].head(10)
```

```python
df[[
    "Car_Age",
    "Kilometers_Per_Year",
    "Owner_Count",
    "Unregistered"
]].describe()
```

```python
print("Missing values:")
print(df[[
    "Car_Age",
    "Kilometers_Per_Year",
    "Owner_Count",
    "Unregistered"
]].isnull().sum())
```

```python
print(df["Owner_Count"].value_counts().sort_index())
```

```python
top_locations = df["Location"].value_counts().head(10).index

df["Location_Grouped"] = df["Location"].where(
    df["Location"].isin(top_locations),
    "Other"
)
```

```python
print(df["Location_Grouped"].value_counts())
```

```python
df.to_csv("../data/processed/car_new_details.csv", index=False)
```
